## Tokens can "see" the text

```python
enc = tiktoken.get_encoding("o200k_base")

def tokens_word(n) -> str:

  """1 токен, 2 токени, 5 токенів - правильне закінчення"""
  if n % 10 ==1 and n % 100 != 11:
    return "токен"
  if n % 10 in (2,3,4) and n % 100 in (12, 13, 14):
    return "токени"
  return "токенів"

def show_tokens(text: str, encoder: tiktoken.core.Encoding = enc)->int:
  ids = encoder.encode(text)
  pieces = [encoder.decode([i]) for i in ids]

  print(f'{len(text)} amount of characters: {len(ids)} {tokens_word(len(ids))}')
  print(ids)
  print(" | ".join(pieces))

  display(pd.DataFrame({
      "ids": ids,
      "pieces": pieces
      }).T.style.hide(axis = "index").hide(axis="columns"))

  return len(ids)

```

```python
text_en = "Large language language models predict next token."
text_uk = "Великі мовні моделі передбачають наступний токен"

n_en = show_tokens(text_en)
print()
n_uk = show_tokens(text_uk)
print()
print(f"Ukranian text needs {n_uk/ n_en:.1f} more tokens than english")
```
